Exercise 3

```python
random = np.random.uniform(low=0.0, high=1, size=(n, 2))

n_init = 0

for i in range(n):
    if random[i,0]**2 + random[i,1]**2 <= 1:
        n_init += 1

print(4 * n_init / n)
```
