```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define probability densities
def p(x):
    return 2 * x

def q(x):
    return np.ones_like(x)

# 2. Generate proposal samples
samples_q = np.random.uniform(0, 1, 10000)

# 3. Evaluate the target density
p_values = p(samples_q)

# 4. Calculate acceptance probabilities
M = 2
accept_probs = p_values / (M * q(samples_q))

# 5. Generate random numbers for acceptance/rejection
random_numbers = np.random.uniform(0, 1, 10000)

# 6. Accept or reject candidates using a Boolean mask
mask = accept_probs >= random_numbers
accepted_samples = samples_q[mask]

# 7. Plot histogram and theoretical PDF
plt.hist(accepted_samples, bins=20, density=True, 
         alpha=0.6, label="Accepted samples")

x_plot = np.linspace(0, 1, 200)
plt.plot(x_plot, p(x_plot), label="Target PDF p(x)")

plt.xlabel("x")
plt.ylabel("Probability density")
plt.legend()
plt.show()

# 8. Calculate acceptance rate
accept_rate = len(accepted_samples) / len(samples_q)

print(f"Proposed samples: {len(samples_q)}")
print(f"Accepted samples: {len(accepted_samples)}")
print(f"Acceptance rate: {accept_rate:.2%}")
print(f"Expected acceptance rate: {1/M:.2%}")
```
